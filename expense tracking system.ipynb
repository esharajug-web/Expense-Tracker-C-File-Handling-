```python
%%writefile student.cpp
#include <iostream>
#include <fstream>
#include <string>
using namespace std;

#include <iostream>
#include <fstream>
#include <string>
using namespace std;

class Expense {
public:
    string category;
    float amount;

    void input() {
        cout << "Enter category (Food/Travel/etc): ";
        cin >> category;
        cout << "Enter amount: ";
        cin >> amount;
    }

    void display() {
        cout << "Category: " << category
             << " | Amount: " << amount << endl;
    }
};

// Add expense
void addExpense() {
    Expense e;
    ofstream file("expenses.txt", ios::app);

    e.input();
    file << e.category << "," << e.amount << endl;

    file.close();
    cout << "Expense added!\n";
}

// View all expenses
void viewExpenses() {
    ifstream file("expenses.txt");
    Expense e;

    cout << "\n--- Expenses ---\n";
    while (getline(file, e.category, ',')) {
        file >> e.amount;
        file.ignore();
        e.display();
    }

    file.close();
}

// Total expense
void totalExpense() {
    ifstream file("expenses.txt");
    Expense e;
    float total = 0;

    while (getline(file, e.category, ',')) {
        file >> e.amount;
        file.ignore();
        total += e.amount;
    }

    cout << "Total Spending: " << total << endl;
    file.close();
}

// Main menu
int main() {
    int choice;

    do {
        cout << "\n===== Expense Tracker =====\n";
        cout << "1. Add Expense\n";
        cout << "2. View Expenses\n";
        cout << "3. Total Spending\n";
        cout << "4. Exit\n";
        cout << "Enter choice: ";
        cin >> choice;

        switch (choice) {
            case 1: addExpense(); break;
            case 2: viewExpenses(); break;
            case 3: totalExpense(); break;
            case 4: cout << "Exiting...\n"; break;
            default: cout << "Invalid choice!\n";
        }

    } while (choice != 4);

    return 0;
}
```
